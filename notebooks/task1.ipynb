ML-PIPELINE

```python
# Categorical Columns
cat_cols = [
    'City',
    'Currency',
    'Has Table booking',
    'Has Online delivery',
    'Is delivering now',
    'Switch to order menu'
]
```

```python
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import OneHotEncoder
from sklearn.pipeline import Pipeline
from sklearn.compose import ColumnTransformer
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score
from sklearn.metrics import mean_squared_error
```

```python
cuisines_pipe=Pipeline([("imp_cuisines",SimpleImputer(strategy="most_frequent")),
                        ("ohe_cuisines",OneHotEncoder(sparse_output=False,drop="first",handle_unknown="ignore"))])
```

```python
preprocessor=ColumnTransformer([("cuisines_imp_ohe",cuisines_pipe,["Cuisines"]),
                             ("cat_ohe",OneHotEncoder(sparse_output=False,drop="first",handle_unknown="ignore"),cat_cols)],remainder="passthrough")
```

#LINEAR_REGRESSION MODEL

```python
pipe_lr=Pipeline([("preprocessor",preprocessor),
                  ("model",LinearRegression())])
```

```python
pipe_lr.fit(x_train,y_train)
```

```python
# Predict
y_pred_lr = pipe_lr.predict(x_test)
```

```python
print(r2_score(y_test,y_pred_lr))
```

```python
print(mean_squared_error(y_test,y_pred_lr))
```

DECISION_TREE_REGRESSION

```python
#using DecisionTreeRegressor
```

```python
from sklearn.tree import DecisionTreeRegressor
```

```python
cuisines_pipe=Pipeline([("imp_cuisines",SimpleImputer(strategy="most_frequent")),
                        ("ohe_cuisines",OneHotEncoder(sparse_output=False,drop="first",handle_unknown="ignore"))])
```

```python
preprocessor=ColumnTransformer([("cuisines_imp_ohe",cuisines_pipe,["Cuisines"]),
                             ("cat_ohe",OneHotEncoder(sparse_output=False,drop="first",handle_unknown="ignore"),cat_cols)],remainder="passthrough")
```

```python
pipe_dt=Pipeline([("preprocessor",preprocessor),
                  ("dt_model",DecisionTreeRegressor(random_state=42))])
```

```python
pipe_dt.fit(x_train,y_train)
```

```python
y_pred_dt=pipe_dt.predict(x_test)

```

```python
print("r2_score:",r2_score(y_test,y_pred_dt))
```

```python
print("MSE :", mean_squared_error(y_test, y_pred_dt))
```

```python
import pandas as pd

results = pd.DataFrame({
    "Model": ["Linear Regression", "Decision Tree Regressor"],
    "R2 Score": [
        r2_score(y_test, y_pred_lr),
        r2_score(y_test, y_pred_dt)
    ],
    "MSE": [
        mean_squared_error(y_test, y_pred_lr),
        mean_squared_error(y_test, y_pred_dt)
    ]
})

results
```

```python
"""Two regression models were trained and evaluated:
Linear Regression and Decision Tree Regressor.

Linear Regression achieved an R² score of 0.3692 and an MSE of 1.3892.

Decision Tree Regressor achieved an R² score of 0.9178 and an MSE of 0.1811.

Since the Decision Tree Regressor has a significantly higher R² score and lower Mean Squared Error, it performs much better on this dataset and was selected as the final model for restaurant rating prediction."""
```
